```python
import requests
def get_place_info(address, api_key):
# Base URL
  base_url = "https://maps.googleapis.com/maps/api/place/findplacefromtext/json"
# Parameters in a dictionary
  params = {
   "input": address,
   "inputtype": "textquery",
   "fields": "formatted_address,name,business_status,place_id",
   "key": api_key,
  }
# Send request and capture response
  response = requests.get(base_url, params=params)
# Check if the request was successful
  if response.status_code == 200:
    return response.json()
  else:
    return None
```

```python
import requests
import json

# Function to get coordinates for a given place name
def get_place_coordinates(place_name):
    url = "https://maps.googleapis.com/maps/api/place/findplacefromtext/json"
    # Define the request parameters. We request the name and geometry (lat/lng).
    params = {
        "input": place_name,
        "inputtype": "textquery",
        "fields": "name,geometry",
        "key": "AIzaSyA2SfFDmJjedIgyRGgLFG5rkP31FQdvzvA"
    }
    
    # Make the request
    response = requests.get(url, params=params)
    result = response.json()
    
    # Print the full JSON response for debugging (optional)
    #print("Full Response:")
    #print(json.dumps(result, indent=2, ensure_ascii=False))
    
    # Check for a successful response and at least one candidate
    if result.get("status") == "OK" and result.get("candidates"):
        candidate = result["candidates"][0]
        if "geometry" in candidate and "location" in candidate["geometry"]:
            lat = candidate["geometry"]["location"]["lat"]
            lng = candidate["geometry"]["location"]["lng"]
            name = candidate.get("name", "No name provided")
            return name, lat, lng
        else:
            print("No geometry information found for this place.")
            return None, None, None
    else:
        print(f"Error or no results found for '{place_name}'. Status: {result.get('status')}")
        return None, None, None
```

```python
import requests
import time

# Function to reverse geocode using latitude and longitude
def get_city_state_country(lat, lng, api_key):
    url = 'https://maps.googleapis.com/maps/api/geocode/json'
    params = {'latlng': f"{lat},{lng}", 'key': api_key}
    try:
        # Request to reverse geocode the lat/lng
        r = requests.get(url, params=params)
        result_dic = r.json()
        
        # Check if the request was successful and if any results were found
        if result_dic['status'] == 'OK' and len(result_dic['results']) > 0:
            city = None
            state = None
            country = None
            for component in result_dic['results'][0]['address_components']:
                if 'locality' in component['types']:  # Locality is generally the city
                    city = component['long_name']
                if 'administrative_area_level_1' in component['types']:  # Administrative area is the state
                    state = component['long_name']
                if 'country' in component['types']:  # Country
                    country = component['long_name']
            
            # Return the city, state, and country if found
            return city, state, country
        else:
            print(f"No results found for lat: {lat}, lng: {lng} - Status: {result_dic['status']}")
            return None, None, None

    except Exception as e:
        print(f"Error while reverse geocoding lat: {lat}, lng: {lng} - {e}")
        return None, None, None

# Main loop
for patent in tqdm(patents_list, desc="Processing patents", unit="patent"):
    lat = patent.get('latitude')
    lng = patent.get('longitude')
    
    if lat is not None and lng is not None:
        # Now reverse geocode the lat/lng to get city, state, and country
        city, state, country = get_city_state_country(lat, lng, api_key)
        patent['applicant_city_name'] = city
        patent['applicant_state_name'] = state
        patent['applicant_country_name'] = country
        # print(f"Patent: {patent['Applicants']}")
        # print(f"Latitude: {lat}")
        # print(f"Longitude: {lng}")
        # print(f"City: {city}")
        # print(f"State: {state}")
        # print(f"Country: {country}")
    else:
        print(f"Latitude and Longitude are missing for patent: {patent['Applicants']}")
    
    # Sleep to avoid hitting API limits
    time.sleep(0.5)
```
